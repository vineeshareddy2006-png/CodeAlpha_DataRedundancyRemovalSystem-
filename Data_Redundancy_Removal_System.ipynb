```python

import sqlite3

# Connect to the database
conn = sqlite3.connect("data_records.db")
cursor = conn.cursor()

# Create a table with a unique email constraint
cursor.execute("""
CREATE TABLE IF NOT EXISTS users (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    email TEXT NOT NULL UNIQUE
)
""")

# Function to validate and insert records
def add_record(name, email):
    name = name.strip()
    email = email.strip().lower()

    if not name or not email or "@" not in email:
        print("Invalid record:", name, email)
        return

    try:
        cursor.execute(
            "INSERT INTO users (name, email) VALUES (?, ?)",
            (name, email)
        )
        conn.commit()
        print("Added successfully:", name, email)

    except sqlite3.IntegrityError:
        print("Duplicate record rejected:", name, email)


# Sample data for testing
records = [
    ("Ravi", "ravi@gmail.com"),
    ("Priya", "priya@gmail.com"),
    ("Ravi", "RAVI@gmail.com"),
    ("Kiran", "kiran@gmail.com"),
    ("Invalid User", "invalid-email"),
    ("Priya", "priya@gmail.com")
]

print("DATA REDUNDANCY REMOVAL SYSTEM")
print("-" * 35)

for name, email in records:
    add_record(name, email)

# Display the final unique records
print("\nFinal unique records:")
cursor.execute("SELECT id, name, email FROM users")

for record in cursor.fetchall():
    print(record)

# Close the database connection
conn.close()

print("\nTask completed successfully!")
```
